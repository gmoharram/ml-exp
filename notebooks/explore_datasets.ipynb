```python
import medmnist
import matplotlib.pyplot as plt
from torchvision import transforms
```

```python
dataset_name = "dermamnist"
image_size = 224
image_resize = 224
root_dir = "/home/ganna/datasets/"
download = False   
```

```python
info = medmnist.INFO[dataset_name]
DataClass = getattr(medmnist, info["python_class"])

test_dataset = DataClass(split="test", size=image_size, root=root_dir, download=download)
```

```python
i = 0
```

```python
i += 1
example_image, example_label = test_dataset[i]
print(example_label)
example_image = example_image.resize((image_resize, image_resize))
plt.imshow(example_image, cmap="gray")
plt.savefig("example_image.png")
```
