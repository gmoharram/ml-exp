### Load timm models

```python
import timm
import torch
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)
```

```python
all_timm_models = timm.list_models()
vit_timm_models = [model for model in all_timm_models if 'vit_' == model[:4]]
convnext_timm_models = [model for model in all_timm_models if 'convnext' in model]
convnext_sizes = set([model.split('_')[1] for model in convnext_timm_models])
```

```python
model = timm.create_model('convnext_tiny', in_chans=3, num_classes=7, drop_rate=0.1, pretrained=False).to(device)
```
